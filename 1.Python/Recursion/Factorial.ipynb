```python
def fact(num):
    if num ==1 :
        return 1 
    else :
        return num*fact(num-1)
    
fact(10)
```

```python
def sum(n) :
    if n == 0 :
        return 1 
    else :
        return n+sum(n-1)
    
print(sum(5))
```

```python
def fec(n):
    if n == 0  :
        return 0
    if n == 1 :
        return 1 
    else :
        return fec(n-1)+fec(n-2)
    
fec(6)
```

```python
# head Recursion  (first call that print)
# print 1 to n 
def printn(n):
    if n == 0 :
        return 
    else :
        printn(n-1)
        print(n)
    
    
printn(5)
```

```python
# tail resursion (first print that call) (it's best , it do not do call back, it is fast )
# print n to 1 

def print1(n):
    if n ==0 :
        return 
    else :
        print(n)
        print1(n-1)

print1(5)
```
