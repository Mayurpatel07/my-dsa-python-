```python
# check weather password is stronger or not 

def is_strong(password):
    if not any(char.isdigit() for char in password):
        return False 
    if len(password) <= 8 :
        return False 
    if not any(char.islower() for char in password):
        return False 
    if not any(char.isupper() for char in password):
        return False 
    else : 
        return True 
    
print(is_strong("Mayur@2105444"))
```

```python
# check that a string is palindrome
def string(s): 
    s = s.lower()
    s = s.replace(" ","")
    print(s)
    if s == s[::-1] :
        return("it is palindrome")
    else :
        return("its not palindrome")
    
string("A man a plan a canal Panama")
    
```

```python
# calculate factorial of number
def fact(n):
    if n ==0 :
        return 1 
    else :
        return n*fact(n-1)
fact(4)
```
