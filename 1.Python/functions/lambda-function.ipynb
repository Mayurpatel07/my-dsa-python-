Lambda functions are small anonymous functions defined using the lambda keyword. They can have any
number of arguments but only one expression. They are commonly used for short operations or as arguments
to higher-order functions.

```python
# normal function 

def sum(a,b):
    return a+b

sum(5,6)
```

```python
# same function in lamda function

# syntax 
# variable = lamda arguments:expression 

sum = lambda a, b: a+b 
sum(5,3) 

```

```python
# normal function
def even(num):
    if num%2==0 :
        return True
even(4)
```

```python
# lambda function 
even = lambda num : num%2==0

even(4)
```

```python
sum  = lambda a : a>5 and a*3 
sum(10)
```
