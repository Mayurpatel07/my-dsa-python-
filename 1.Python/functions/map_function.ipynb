map()
The map() function applies a given function to all items in an input list (or any other iterable) and returns a map
object (an iterator). This is particularly useful for transforming data in a list comprehensively.

```python
# syntax 
# map(function,ittrable)
```

```python
# using map in normal function 
def sq(x):
    x = x**2
    return x 

nums = [1,2,3,4,5,6,7,8]

list(map(sq,nums))
```

```python
# using map in lambda function 
nums1 = [1,2,3,4,5,6]
list(map(lambda x :x**2 , nums1))
```

```python
# using map in multiple ittrate 

# que: adding each elemts with each element  like 1+11 , 2+22 etc

numbers1 = [1,2,3,4,5]
numbers2 = [11,22,33,44,55]

added_num = list(map(lambda a,b: a+b ,numbers1,numbers2))
print(added_num)

```

```python
# using map for inbuild functions 

# que convert all list string into int 

list1 = ["1",'2','3','4']

int_list = list(map(int,list1))
print(int_list)
```

```python
# take input as list 

list2 = list(map(int,input("enter numbers by space").split()))
print(list2)
```
