A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and
improving readability.

```python
# creating function 
def sum(a,b):
    sum = a+b
    print(sum)

#calling function
sum(5,6) 
```

```python
#default function 
def hello(name = "developer") :  # we give default value in argument 
    print(f"hello {name}")

hello()
```

```python
# positional argument 
def sum(a,b):
    sum = a+b
    print(sum)

#calling function
sum(5,6)  # we are giving 5 as (a) 6 as (b) this are positions 
```

```python
# variable length function 
def nums(*args): # by adding * we can give multiple arguemtns 
    for nums in args:
        print(nums)

nums(1,2,3,3,4,5,6) 
```

```python
# keyword argument 
def info(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

# calling function 
info(name = 'mayur',id= 0) # all the arguments are in key value pairs 
```

```python
# return statement 
def multi(a):
    return a*2 # it will return value 

multi(5)
```
