```python
# Encapsulation
# Encapsulation means hiding internal details of a class and only exposing what’s necessary.
# It helps to protect important data from being changed directly and keeps the code secure and organized.
# using private and protected access modifier 

class employee :
    def __init__(self,name,salary,role) :
        self.name=name 
        self.__salary=salary  #using double underscore for private access modifier
        self.role=role
    def show_salary(self):
        print(f"{self.name} is working for {self.role} and has {self.__salary}")

employee1 = employee("mayur",50000,"SD1")
employee1.show_salary()
```
