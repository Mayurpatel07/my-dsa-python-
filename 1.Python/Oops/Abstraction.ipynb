```python
# Abstraction 
# hiding unnecesary deatils from users through classes and method
class student :
    def __init__(self,first_name,std,percent) :   #creatimg method
        self.name = first_name            #Attribute
        self.std = std 
        self.percent=percent
    def marks(self):                     #abstraction 
        print(f"{self.name} scored {self.percent}% in class {self.std}")

stu = student("mayur",10,88)  #Creating Object 

stu.marks()
```
