```python
# creating class
class student :
    def __init__(self,first_name,age) :   #creatimg method
        self.name = first_name            #Attribute
        self.age = age 

std = student("mayur",20)  #Creating Object 

print(std.name) 
```

```python
class dog :
    def __init__(self,name,age):
        self.name = name 
        self.age = age 
    def bark(self):

        print(f"{self.name} is barking")

dog1 = dog("yash",11)
dog1 = dog("vedant",10)

dog1.bark()
```
