```python
def insertion_sort(nums):
    n = len(nums)
    for i in range(1,n):
        key = i 
        j=i -1 
        while j>=0 :
            if nums[j]>key:
                nums[j+1] = nums[j]
                j = j-1
            else :
                break
            nums[j+1] = key 
    return nums

arr = [10,2,5,8,1,3,2]

print(insertion_sort(arr))
```
