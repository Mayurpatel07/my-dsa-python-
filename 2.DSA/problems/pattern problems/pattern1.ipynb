```python
# Pyramid Pattern using *
n = int(input("Enter num"))
for i in range(n):
    for j in range(n-i-1):
        print(" ",end="")
    for k in range(i+1):
        print("*",end=" ")
    print()
```

```python
# Inverted Pyramid Pattern using *
n = int(input("enter n"))
for i in range(n):
    for j in range(i):
        print(" ",end="")
    for k in range(n-i):
        print("*",end=" ")
    print()
```

```python
# Given an integer N, print a right-angled triangle using numbers
n = int(input("enter number"))
for i in range(n):
    for j in range(i+1):
        print(j+1,end=" ")
    print()
```

```python
# print number in continuse form using right angle trangle (take n from user)
n = int(input("Enter Number"))
num = 1 
for i in range(n):
    for j in range(i+1):
        print(num,end=" ")
        num = num +1
    print()
```

```python
# print diamond usinf *
n = int(input("enter number"))

for i in range(n-1):
    for j in range(n-i-1):
        print(" ",end="")
    for k in range(i+1):
        print("*",end=" ")
    print()
for i in range(n):
    for j in range(i):
        print(" ",end="")
    for k in range(n-i):
        print("*",end=" ")
    print()
```

```python
# print Sandglass Pattern using *
n = int(input("enter number"))
for i in range(n-1):
    for j in range(i):
        print(" ",end="")
    for k in range(n-i):
        print("* ",end="")
    print()
for i in range(n):
    for j in range(n-i-1):
        print(' ',end="")
    for k in range(i+1):
        print("* ",end="")
    print()
```

```python
# Hollow Right Triangle
n = int(input("enter number"))

for i in range(n):
    for j in range(i+1):
        if j==0 or i==n-1 or i==j :
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
