```python
def binarySearch(num,key) :
    n = len(num)-1
    start = 0 
    end = n
    while start <= end :
        mid = (start +end) //2 
        if num[mid] == key :
            return mid
        elif num[mid] < key :
            start = mid + 1
        else:
            end = mid +1
        
    return -1 
        
num = [1,2,3,4,5,6,7,8]
key = 5

result = binarySearch(num,key)
print(result)
    

    
```

```python
def bs(nums,target):

    start = 0 
    end = len(nums)-1
    
    while start <= end :
        mid = (start + end)//2
        if nums[mid] == target :
            return mid 
        elif nums[mid] < target:
            start = mid +1 
        else :
            end = mid -1 
    return -1 


nums =[1,2,3,4,5]
target = 4
print(bs(nums,target))
```
